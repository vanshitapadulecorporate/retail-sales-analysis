```python
import pandas as pd
import random
from datetime import datetime, timedelta
```

```python
categories = {
    "Mobile": {
        "Smartphone": [
            ("Samsung", "Galaxy A56"),
            ("Apple", "iPhone 15"),
            ("OnePlus", "Nord CE4"),
            ("Realme", "Narzo 70"),
            ("Redmi", "Note 13")
        ]
    },

    "Laptop": {
        "Gaming Laptop": [
            ("Asus", "ROG Strix"),
            ("Acer", "Nitro V"),
            ("Lenovo", "Legion 5")
        ],

        "Business Laptop": [
            ("HP", "Pavilion 14"),
            ("Dell", "Inspiron 15")
        ]
    },

    "Fashion": {
        "Shoes": [
            ("Nike", "Air Max"),
            ("Adidas", "Ultraboost"),
            ("Puma", "Runner X")
        ],

        "Jeans": [
            ("Levis", "511 Slim"),
            ("Wrangler", "Classic Fit")
        ]
    },

    "Electronics": {
        "Headphones": [
            ("Boat", "Rockerz 450"),
            ("Sony", "WH-CH520"),
            ("JBL", "Tune 760")
        ],

        "Smart Watch": [
            ("Noise", "ColorFit"),
            ("Fire-Boltt", "Ninja"),
            ("Boat", "Wave Neo")
        ]
    }
}

suppliers = [
    "Tech World Pvt Ltd",
    "Digital Hub",
    "Elite Fashion",
    "Future Retail",
    "Smart Electronics"
]

warranty = [
    "6 Months",
    "12 Months",
    "24 Months"
]

status = [
    "Active",
    "Discontinued"
]
colors = [
    "Black",
    "White",
    "Blue",
    "Red",
    "Silver",
    "Gray",
    "Green"
]

gst = [5, 12, 18, 28]

return_eligible = [
    "Yes",
    "No"
]
```

```python
products = []

for i in range(1, 101):

    category = random.choice(list(categories.keys()))

    subcategory = random.choice(list(categories[category].keys()))

    brand, model = random.choice(categories[category][subcategory])

    product_name = brand + " " + model

    
    if category == "Mobile":
        cost_price = random.randint(8000, 60000)

    elif category == "Laptop":
        cost_price = random.randint(25000, 90000)

    elif category == "Fashion":
        cost_price = random.randint(500, 5000)

    else:
        cost_price = random.randint(800, 20000)

   
    profit = random.randint(500, 8000)

   
    selling_price = cost_price + profit

   
    discount = random.choice([5, 10, 15, 20, 25, 30, 35])

 
    stock = random.randint(10, 500)

    
    rating = round(random.uniform(3.5, 5.0), 1)

   
    supplier = random.choice(suppliers)

    
    launch_date = (
        datetime(2024, 1, 1)
        + timedelta(days=random.randint(0, 730)) #random number of days within approximately three years timedelta add those days to the starting date
    ).date()

    

    color = random.choice(colors)

    gst_value = random.choice(gst)

    return_status = random.choice(return_eligible)

   
    product = [
        f"P{i:03}",
        product_name,
        category,
        subcategory,
        brand,
        cost_price,
        selling_price,
        profit,
        discount,
        stock,
        rating,
        supplier,
        launch_date,
        random.choice(warranty),
        random.choice(status),
        color,
        gst_value,
        return_status
    ]

    products.append(product)
```

```python
products_df = pd.DataFrame(
    products,
    columns=[
        "ProductID",
        "ProductName",
        "Category",
        "SubCategory",
        "Brand",
        "CostPrice",
        "SellingPrice",
        "Profit",
        "Discount",
        "StockQuantity",
        "Rating",
        "Supplier",
        "LaunchDate",
        "Warranty",
        "ProductStatus",
        "Color",
        "Gst",
        "ReturnEligible"
    ]
)
```

```python
orders=[]
```

```python
order_status = [
    "Delivered",
    "Cancelled",
    "Returned"
]

sales_channel = [
    "Website",
    "Mobile App"
]

delivery_type = [
    "Standard",
    "Express"
]

coupon_codes = [
    "NONE",
    "SAVE10",
    "WELCOME5",
    "FEST20",
    "NEWUSER"
]

warehouses = [
    "Pune",
    "Mumbai",
    "Delhi",
    "Bengaluru",
    "Hyderabad"
]

priorities = [
    "Low",
    "Medium",
    "High"
]
```

```python
for i in range(1, 10001):

    order_id = f"O{i:06}"

    invoice_no = f"INV{i:06}"

    order_date = (
        datetime(2024, 1, 1) +
        timedelta(days=random.randint(0, 730))
    ).date()

    customer = customers_df.sample(1).iloc[0]

    product = products_df.sample(1).iloc[0]

    customer_id = customer["CustomerID"]

    product_id = product["ProductID"]

    quantity = random.randint(1, 5)

    unit_price = product["SellingPrice"]

    cost_price = product["CostPrice"]

    discount = random.choice([0, 5, 10, 15, 20])

    gst = product["Gst"]

    payment = customer["PreferredPayment"]

    sales_channel_value = random.choice(sales_channel)

    warehouse = random.choice(warehouses)

    delivery = random.choice(delivery_type)

    shipping_charge = random.choice([0, 49, 99, 149])

    coupon = random.choice(coupon_codes)

    status = random.choices(
        ["Delivered", "Cancelled", "Returned"],
        weights=[85, 10, 5]
    )[0]

    priority = random.choice(priorities)

    festival_sale = random.choice(["Yes", "No"])

    customer_rating = random.randint(1, 5)

    return_status = "Yes" if status == "Returned" else "No"

    sales_amount = quantity * unit_price

    discount_amount = sales_amount * discount / 100

    taxable_amount = sales_amount - discount_amount

    gst_amount = taxable_amount * gst / 100

    final_sales = taxable_amount + gst_amount

    profit = (unit_price - cost_price) * quantity

    hour = random.randint(8, 22)

    if hour < 12:
        time_of_day = "Morning"
    elif hour < 17:
        time_of_day = "Afternoon"
    else:
        time_of_day = "Evening"

    month = order_date.strftime("%B")

    quarter = f"Q{((order_date.month-1)//3)+1}"

    year = order_date.year

    salesperson = f"EMP{random.randint(1,50):03}"

    orders.append([
        order_id,
        invoice_no,
        order_date,
        customer_id,
        product_id,
        quantity,
        unit_price,
        cost_price,
        discount,
        gst,
        round(final_sales,2),
        round(profit,2),
        payment,
        status,
        coupon,
        festival_sale,
        sales_channel_value,
        warehouse,
        delivery,
        shipping_charge,
        customer_rating,
        return_status,
        priority,
        salesperson,
        time_of_day,
        month,
        quarter,
        year
    ])
```

```python
orders_df = pd.DataFrame(
    orders,
    columns=[
        "OrderID",
        "InvoiceNo",
        "OrderDate",
        "CustomerID",
        "ProductID",
        "Quantity",
        "UnitPrice",
        "CostPrice",
        "Discount",
        "GST",
        "SalesAmount",
        "Profit",
        "PaymentMethod",
        "OrderStatus",
        "CouponCode",
        "FestivalSale",
        "SalesChannel",
        "Warehouse",
        "DeliveryType",
        "ShippingCharge",
        "CustomerRating",
        "ReturnStatus",
        "OrderPriority",
        "SalesPersonID",
        "TimeOfDay",
        "Month",
        "Quarter",
        "Year"
    ]
)
```

```python
customer_lookup = customers_df.set_index("CustomerID")
```

```python
shipping = []

couriers = [
    "Delivery",
    "Blue Dart",
    "DTDC",
    "Ecom Express",
    "XpressBees",
    "India Post"
]

statuses = [
    "Delivered",
    "In Process",
    "Out for Delivery",
    "Delayed"
]
```

```python
for i, order in orders_df.iterrows():

    shipping_id = f"SH{i+1:06}"

    order_id = order["OrderID"]

    customer = customer_lookup.loc[order["CustomerID"]]

    courier = random.choice(couriers)

    warehouse = order["Warehouse"]

    dispatch_date = order["OrderDate"] + timedelta(days=random.randint(1,2))

    estimated_delivery = dispatch_date + timedelta(days=random.randint(2,5))

    actual_delivery = estimated_delivery + timedelta(days=random.randint(-1,2))

    if actual_delivery <= estimated_delivery:
        shipping_status = "Delivered"
    else:
        shipping_status = random.choice(statuses)

    delivery_days = (actual_delivery - dispatch_date).days

    shipping_cost = random.choice([0,49,99,149,199])

    tracking_number = f"TRK{random.randint(100000,999999)}"

    delivery_city = customer["City"]

    delivery_state = customer["State"]

    delivery_zone = customer["Region"]

    delivery_attempt = random.randint(1,3)

    customer_received = "Yes" if shipping_status == "Delivered" else "No"

    shipping.append([
        shipping_id,
        order_id,
        courier,
        warehouse,
        dispatch_date,
        estimated_delivery,
        actual_delivery,
        shipping_status,
        delivery_days,
        shipping_cost,
        tracking_number,
        delivery_city,
        delivery_state,
        delivery_zone,
        delivery_attempt,
        customer_received
    ])
```

```python
shipping_df = pd.DataFrame(
    shipping,
    columns=[
        "ShippingID",
        "OrderID",
        "CourierPartner",
        "Warehouse",
        "DispatchDate",
        "EstimatedDelivery",
        "ActualDelivery",
        "ShippingStatus",
        "DeliveryDays",
        "ShippingCost",
        "TrackingNumber",
        "DeliveryCity",
        "DeliveryState",
        "DeliveryZone",
        "DeliveryAttempt",
        "CustomerReceived"
    ]
)
```

```python
review_orders = orders_df.sample(
    n=8000,
    random_state=42
).reset_index(drop=True)
```

```python
reviews = []

titles = [
    "Excellent Product",
    "Very Good",
    "Worth the Money",
    "Average Quality",
    "Not Satisfied",
    "Highly Recommended",
    "Good Value",
    "Amazing Product"
]

texts = [
    "Product quality is excellent.",
    "Fast delivery and good packaging.",
    "Value for money.",
    "Satisfied with the purchase.",
    "Could be better.",
    "Highly recommended.",
    "Met my expectations.",
    "Will buy again.",
    "Excellent customer service.",
    "Packaging was damaged but product is good."
]
```

```python
for i, order in review_orders.iterrows():

    review_id = f"R{i+1:06}"

    rating = random.choices(
        [5,4,3,2,1],
        weights=[40,30,15,10,5]
    )[0]

    review_date = order["OrderDate"] + timedelta(
        days=random.randint(3,15)
    )

    recommend = "Yes" if rating >=4 else "No"

    helpful = random.randint(0,100)

    reviews.append([
        review_id,
        order["OrderID"],
        order["CustomerID"],
        order["ProductID"],
        rating,
        random.choice(titles),
        random.choice(texts),
        review_date,
        recommend,
        "Yes",
        helpful
    ])
```

```python
reviews_df = pd.DataFrame(
    reviews,
    columns=[
        "ReviewID",
        "OrderID",
        "CustomerID",
        "ProductID",
        "Rating",
        "ReviewTitle",
        "ReviewText",
        "ReviewDate",
        "WouldRecommend",
        "VerifiedPurchase",
        "HelpfulVotes"
    ]
)
```
